```python


filename = 'data/cr_2144/F104.csv'

stab_der = StabilityDerivatives()
stab_der.read_csv(filename)
stab_der.body_to_stab()



# CL/CD alpha derivatives
image = plt.imread('data/cr_2144/screenshots/f104/CLa_CDa.png')
transform_CLa = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 6.0))
transform_CDa = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 1.2))
fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cz_alpha', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_CLa, color = color, label = label)
    stab_der.plot_derivative('Cx_alpha', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_CDa, color = color, label = '')
ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')


# Cm alpha, pitch rate, and alpha dot derivatives
image = plt.imread('data/cr_2144/screenshots/f104/Cma_Cmq.png')
transform_Cma = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -3.0))
transform_Cmq = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -12))
fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cm_alpha', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Cma, color = color, label = label)
    stab_der.plot_derivative('Cm_q', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cmq, color = color, label = '')
    stab_der.plot_derivative('Cm_alpha_dot', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cmq, color = color, label = '')
    
ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# CL delta stab and Cm delta stab
image = plt.imread('data/cr_2144/screenshots/f104/CLds_Cmds.png')
transform_CLde = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 1.2))
transform_Cmde = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -3))
fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cz_de', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_CLde, color = color, label = label)
    stab_der.plot_derivative('Cm_de', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cmde, color = color, label = '')

ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# Cy alpha, pitch rate, and alpha dot derivatives
image = plt.imread('data/cr_2144/screenshots/f104/Cyb_Cnb_Clb.png')
transform_Cyb = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -2.0))
transform_Cnb = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.6))
transform_Clb = make_image_axis_transform(image, color="green",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -0.3))

fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cy_beta', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Cyb, color = color, label = label)
    stab_der.plot_derivative('Cn_beta', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cnb, color = color, label = '')
    stab_der.plot_derivative('Cl_beta', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Clb, color = color, label = '')
ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# Cl and Cn due to roll rate
image = plt.imread('data/cr_2144/screenshots/f104/Clp_Cnp.png')
transform_Clp = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -0.5))
transform_Cnp = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -0.3))

fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cl_p', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Clp, color = color, label = label)
    stab_der.plot_derivative('Cn_p', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cnp, color = color, label = '')

ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# Cl and Cn due to yaw rate
image = plt.imread('data/cr_2144/screenshots/f104/Clr_Cnr.png')
transform_Clr = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.4))
transform_Cnr = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, -1.4))

fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cl_r', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Clr, color = color, label = label)
    stab_der.plot_derivative('Cn_r', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cnr, color = color, label = '')

ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# Cl and Cn due to aileron
image = plt.imread('data/cr_2144/screenshots/f104/Clda_Cnda.png')
transform_Clda = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.06))
transform_Cnda = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.02))

fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cl_da', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Clda, color = color, label = label)
    stab_der.plot_derivative('Cn_da', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cnda, color = color, label = '')

ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')

# Cy, Cn, and Cl due to rudder
image = plt.imread('data/cr_2144/screenshots/f104/Cydr_Cndr_Cldr.png')
transform_Cydr_Cndr = make_image_axis_transform(image, color="red",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.2))
transform_Cldr = make_image_axis_transform(image, color="blue",
                                          origin_value=(0.0, 0.0),
                                          x_axis_value=(2.0, 0.0),
                                          y_axis_value=(0.0, 0.04))

fig, ax = plt.subplots(1, 1, figsize = (12, 8))
ax.imshow(image)

for inda, alt in enumerate(np.unique(stab_der.dataframe.Altitude)):
    color = 'C%i' % inda
    label = '%.0f ft' % alt
    stab_der.plot_derivative('Cl_dr', ax = ax, filters = {'Altitude': alt},
                                 linestyle = '', transform = transform_Cldr, color = color, label = label)
    stab_der.plot_derivative('Cn_dr', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cydr_Cndr, color = color, label = '')
    stab_der.plot_derivative('Cy_dr', ax = ax, filters = {'Altitude': alt},
                             linestyle = '', transform = transform_Cydr_Cndr, color = color, label = '')

ax.legend(loc = 'upper left', framealpha = 1.)
ax.axis('off')
ax.set_title('')
```
